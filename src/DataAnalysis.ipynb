```python
import pandas as pd
#dataset from www.kaggle.com/datasets/solomonameh/spotify-music-dataset?resource=download
popular_songs = pd.read_csv("Songs/high_popularity_spotify_data.csv")
print(popular_songs.head())
```

```python
#see variations in scores
import matplotlib.pyplot as plt
import numpy as np
plt.figure(figsize=(6, 4))
plt.scatter(popular_songs['tempo'], popular_songs['energy'], alpha=0.6, edgecolor='k')

plt.xlabel("Tempo (BPM)")
plt.ylabel("Energy")
plt.title("Tempo vs Energy for All Popular Spotify Songs")

plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(popular_songs['tempo'], popular_songs['danceability'], alpha=0.6, edgecolor='k')

plt.xlabel("Tempo (BPM)")
plt.ylabel("Danceability")
plt.title("Tempo vs Danceability for All Popular Spotify Songs")

plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(popular_songs['energy'], popular_songs['danceability'], alpha=0.6, edgecolor='k')

plt.xlabel("Energy")
plt.ylabel("Danceability")
plt.title("Energy vs Danceability for All Popular Spotify Songs")

plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
```

```python
#binary classifier input song A and input song B snip, see if the transition is "good"
skrillex_features = pd.read_csv("DJ_Sets/skrillex2015_features.csv")
skrillex_transitions = pd.read_csv("DJ_Sets/skrillex2015_transitions.csv")
print("import complete")
```

```python
skrillex_transitions.dtypes
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(skrillex_transitions['delta_tempo'], skrillex_transitions['delta_rms'], alpha=0.6, edgecolor='k')

plt.xlabel("Tempo Change")
plt.ylabel("Energy Change")
plt.title("Tempo Change vs Energy Change for Skrillex Set Transitions")

plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(skrillex_transitions['delta_tempo'], skrillex_transitions['delta_flux'], alpha=0.6, edgecolor='k')

plt.xlabel("Tempo Change")
plt.ylabel("Spectral Flux")
plt.title("Tempo Change vs Spectral Flux for Skrillex Set Transitions")

plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
#looks like tempo vs energy changes with an inversion across X-axis (tempo)
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(skrillex_transitions['delta_rms'], skrillex_transitions['delta_flux'], alpha=0.6, edgecolor='k')

plt.xlabel("Energy Change")
plt.ylabel("Flux")
plt.title("Energy Change vs Flux for Skrillex Set Transitions")

plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
#looks like tempo vs energy changes with an inversion across X-axis (tempo)
```

```python
import matplotlib.pyplot as plt
import numpy as np

i = 0
mfcc_cols = [f"delta_mfcc_{k}" for k in range(13)]
delta = skrillex_transitions.loc[i, mfcc_cols].values

plt.stem(np.arange(13), delta)
plt.title(f"MFCC Delta for Transition {i}")
plt.xlabel("Coefficient index")
plt.ylabel("Delta MFCC")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
mfcc_cols = [f"delta_mfcc_{k}" for k in range(13)]
matrix = skrillex_transitions[mfcc_cols]

plt.figure(figsize=(12, 6))
sns.heatmap(matrix, cmap="coolwarm", xticklabels=mfcc_cols, yticklabels=skrillex_transitions.index)
plt.title("MFCC Delta Heatmap for All Transitions")
plt.xlabel("MFCC Coefficient")
plt.ylabel("Transition Index")
plt.tight_layout()
plt.show()
```
